```python
import json
import matplotlib.pyplot as plt
import numpy as np

# ----------------------------
# Configuration
# ----------------------------

files = [
    "old/eval_linucb.json",
    "old/eval_ts.json",
    "old/eval_dqn.json",
    "old/eval_dqn_ablation.json",
    "old/eval_static_threshold.json",
    "old/eval_static_tiers.json"
]

model_names = [
    "LinUCB",
    "Thompson Sampling",
    "DQN",
    "DQN without H-Score",
    "Static Single Threshold",
    "Static Tiered Threshold"
]

plt.rcParams.update({
    "font.size": 11,
    "axes.titlesize": 11,
    "axes.labelsize": 12,
    "legend.fontsize": 10,
    "xtick.labelsize": 8,
    "ytick.labelsize": 8,
    "lines.linewidth": 2.8,
    "axes.linewidth": 0.3
})

# ----------------------------
# Load data
# ----------------------------

all_data = []
max_cols = 0

global_min_t = float("inf")
global_max_t = float("-inf")

for file in files:
    with open(file, "r") as f:
        data = json.load(f)
        all_data.append(data)
        max_cols = max(max_cols, len(data.keys()))

        for sim in data.values():
            t = sim["timesteps"]
            global_min_t = min(global_min_t, min(t))
            global_max_t = max(global_max_t, max(t))

n_models = len(all_data)
n_cols = max_cols

# ----------------------------
# Compute column-wise max Y
# ----------------------------

column_max_y = []

for col in range(n_cols):
    col_max = 0
    for model_data in all_data:
        keys = list(model_data.keys())
        if col < len(keys):
            sim = keys[col]
            dct = model_data[sim]
            col_max = max(
                col_max,
                max(dct["human_counts"]),
                max(dct["bot_counts"])
            )
    column_max_y.append(col_max)

# ----------------------------
# Define X ticks (global)
# ----------------------------

num_xticks = 5
xticks = np.linspace(global_min_t, global_max_t, num_xticks)
xticks = np.round(xticks).astype(int)

# ----------------------------
# Create grid
# ----------------------------

fig, axes = plt.subplots(
    n_models,
    n_cols,
    figsize=(3.5 * n_cols, 3.5 * n_models),
    sharex=True,
    sharey=False   # important change
)

for ax in axes.flat:
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

if n_models == 1:
    axes = np.expand_dims(axes, axis=0)
if n_cols == 1:
    axes = np.expand_dims(axes, axis=1)

# ----------------------------
# Plot
# ----------------------------

for i, (model_data, model_name) in enumerate(zip(all_data, model_names)):

    keys = list(model_data.keys())

    for j in range(n_cols):

        ax = axes[i, j]

        if j >= len(keys):
            ax.axis("off")
            continue

        sim = keys[j]
        dct = model_data[sim]

        timesteps = dct["timesteps"]
        humans = dct["human_counts"]
        bots = dct["bot_counts"]

        ax.plot(timesteps, humans,
                color="tab:blue",
                linewidth=1.8,
                label="Humans",
                zorder=3
                )

        ax.plot(timesteps, bots,
                color="tab:red",
                linestyle="--",
                linewidth=1.8,
                label="Bots",
                zorder=3
                )

        # Set consistent X across all
        ax.set_xlim(global_min_t, global_max_t)
        ax.set_xticks(xticks)
        ax.tick_params(axis='x', labelbottom=True)

        # Column-wise Y scaling
        ymax = column_max_y[j]
        ax.set_ylim(0, ymax)

        ax.set_ylim(top=ymax * 1.02)

        # ax.margins(y=0.2)

        # Column-wise Y ticks
        yticks = np.linspace(0, ymax, 5)
        yticks = np.round(yticks).astype(int)
        ax.set_yticks(yticks)
        ax.tick_params(axis='y', labelleft=True)

        ax.grid(alpha=0.3, zorder=0)

        # Column titles (top row only)
        if i == 0:
            ax.set_title(sim.replace("_", " ").title() + " Bots", pad=8)

        # Row labels (first column only)
        if j == 0:
            ax.set_ylabel(f"{model_name}\n\nUsers")

        # X label bottom row only
        if i == n_models - 1:
            ax.set_xlabel("Timestep")

# ----------------------------
# Global legend
# ----------------------------

handles, labels = axes[0, 0].get_legend_handles_labels()
fig.legend(
    handles,
    labels,
    loc="upper center",
    ncol=2,
    frameon=False,
    bbox_to_anchor=(0.5, 0.98)
)

# ----------------------------
# Layout + Save
# ----------------------------

plt.subplots_adjust(
    left=0.07,
    right=0.98,
    top=0.94,
    bottom=0.06,
    wspace=0.15,
    hspace=0.20
)

plt.savefig("population_dynamics_grid.pdf", dpi=300)
plt.show()
```

```python
import json
import matplotlib.pyplot as plt
import numpy as np

# ----------------------------
# Configuration
# ----------------------------

files = [
    # "old/eval_linucb.json",
    "old/eval_linucb_new1.json",
    "old/eval_ts.json",
    "old/eval_dqn.json",
    "old/eval_dqn_ablation.json",
    "old/eval_static_threshold.json",
    "old/eval_static_tiers.json"
]

model_names = [
    "LinUCB",
    "Thompson Sampling",
    "DQN",
    "DQN without H-Score",
    "Static Single-Threshold",
    "Static Multi-Threshold"
]

plt.rcParams.update({
    "font.size": 11,
    "axes.titlesize": 14,
    "axes.labelsize": 12,
    "legend.fontsize": 18,
    "xtick.labelsize": 8,
    "ytick.labelsize": 8,
    "lines.linewidth": 2.8,
    "axes.linewidth": 0.3
})

# ----------------------------
# Load data
# ----------------------------

all_data = []
max_rows = 0

global_min_t = float("inf")
global_max_t = float("-inf")

for file in files:
    with open(file, "r") as f:
        data = json.load(f)
        all_data.append(data)
        max_rows = max(max_rows, len(data.keys()))

        for sim in data.values():
            t = sim["timesteps"]
            global_min_t = min(global_min_t, min(t))
            global_max_t = max(global_max_t, max(t))

n_models = len(all_data)
n_rows = max_rows  # simulations
n_cols = n_models  # models

# ----------------------------
# Compute row-wise max Y (per simulation)
# ----------------------------

row_max_y = []

for row in range(n_rows):
    row_max = 0
    for model_data in all_data:
        keys = list(model_data.keys())
        if row < len(keys):
            sim = keys[row]
            dct = model_data[sim]
            row_max = max(
                row_max,
                max(dct["human_counts"]),
                max(dct["bot_counts"])
            )
    row_max_y.append(row_max)

# ----------------------------
# Define X ticks (global)
# ----------------------------

num_xticks = 5
xticks = np.linspace(global_min_t, global_max_t, num_xticks)
xticks = np.round(xticks).astype(int)

# ----------------------------
# Create grid
# ----------------------------

fig, axes = plt.subplots(
    n_rows,              # rows = simulations
    n_cols,              # cols = models
    figsize=(3.5 * n_cols, 3.5 * n_rows),
    sharex=True,
    sharey=False
)

for ax in axes.flat:
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

if n_rows == 1:
    axes = np.expand_dims(axes, axis=0)
if n_cols == 1:
    axes = np.expand_dims(axes, axis=1)

# ----------------------------
# Plot
# ----------------------------

for row in range(n_rows):  # simulations (rows)

    for col, (model_data, model_name) in enumerate(zip(all_data, model_names)):

        ax = axes[row, col]
        keys = list(model_data.keys())

        if row >= len(keys):
            ax.axis("off")
            continue

        sim = keys[row]
        dct = model_data[sim]

        timesteps = dct["timesteps"]
        humans = dct["human_counts"]
        bots = dct["bot_counts"]

        ax.plot(
            timesteps,
            humans,
            color="tab:blue",
            linewidth=1.8,
            label="Humans",
            zorder=3
        )

        ax.plot(
            timesteps,
            bots,
            color="tab:red",
            linestyle="--",
            linewidth=1.8,
            label="Bots",
            zorder=3
        )

        # Global X scaling
        ax.set_xlim(global_min_t, global_max_t)
        ax.set_xticks(xticks)
        ax.tick_params(axis='x', labelbottom=True)

        # Row-wise Y scaling (per simulation)
        ymax = row_max_y[row]
        ax.set_ylim(0, ymax * 1.02)

        yticks = np.linspace(0, ymax, 5)
        yticks = np.round(yticks).astype(int)
        ax.set_yticks(yticks)
        ax.tick_params(axis='y', labelleft=True)

        ax.grid(alpha=0.3, zorder=0)

        # Column titles (top row only) → model names
        if row == 0:
            ax.set_title(model_name, pad=8)

        # Row labels (first column only) → simulation names
        if col == 0:
            ax.set_ylabel(
                sim.replace("_", " ").title() + " Bots\n\nUsers",
                fontsize=14
            )

        # X label (bottom row only)
        if row == n_rows - 1:
            ax.set_xlabel("Timestep")

# ----------------------------
# Global legend
# ----------------------------

handles, labels = axes[0, 0].get_legend_handles_labels()
fig.legend(
    handles,
    labels,
    loc="upper center",
    ncol=2,
    frameon=False,
    bbox_to_anchor=(0.5, 0.98)
)

# ----------------------------
# Layout + Save
# ----------------------------

plt.subplots_adjust(
    left=0.07,
    right=0.98,
    top=0.94,
    bottom=0.06,
    wspace=0.15,
    hspace=0.20
)

# plt.savefig("population_dynamics_grid.pdf", dpi=300)
plt.show()
```
